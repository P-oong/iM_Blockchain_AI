# 직접 finclassing (부분 보류)

```python
def fine_classing_balanced_bad_rate(data, variable, target, bins=10):
    data = data.sort_values(variable)
    total_bad = data[target].sum()
    total_good = len(data) - total_bad
    bad_per_bin = total_bad / bins
    good_per_bin = total_good / bins
    
    bin_indices = [int(i * len(data) / bins) for i in range(1, bins)]
    bin_thresholds = [data[variable].iloc[idx] for idx in bin_indices]
    
    bin_thresholds = [-np.inf] + bin_thresholds + [np.inf]
    
    data['Bin'] = pd.cut(data[variable], bins=bin_thresholds, labels=range(bins))
    return data

# KS 통계량 계산 함수
def calculate_ks(data, variable, target):
    grouped = data.groupby(variable, as_index=False)
    ks_statistic, _ = stats.ks_2samp(grouped.get_group(0)[target], grouped.get_group(1)[target])
    return ks_statistic

# t-검정
def perform_t_test(data, variable, target):
    grouped = data.groupby(variable, as_index=False)
    t_statistic, p_value = stats.ttest_ind(grouped.get_group(0)[target], grouped.get_group(1)[target])
    return t_statistic, p_value


def plot_bin_line(data, variable, target):
    grouped = data.groupby('Bin', as_index=False)
    bin_means = grouped.mean()[[variable, target]]
    
    plt.figure(figsize=(10, 6))
    plt.plot(bin_means[variable], bin_means[target], marker='o')
    
    for i, (x, y) in enumerate(zip(bin_means[variable], bin_means[target])):
        bin_value = i + 1
        plt.annotate(f'Bin {bin_value}', (x, y), textcoords="offset points", xytext=(0,10), ha='center')
    
    plt.xlabel(variable)
    plt.ylabel(target + ' Mean')
    plt.title(f'Distribution of {target} by {variable} Bin')
    plt.grid()
    plt.show()

    
    
# 각 독립변수에 대한 계급 세분화, 시각화, KS 통계량, t-검정 수행
for var_name in df.columns:
    if var_name not in ['년도', '분기', '행정구', '행정동', '지방소멸위험지수','위험여부']:
        data = fine_classing_balanced_bad_rate(df, var_name, '위험여부')
        plot_bin_line(data, var_name, '위험여부')

        
        ks = calculate_ks(data, 'Bin', '위험여부')
        t_statistic, p_value = perform_t_test(data, 'Bin', '위험여부')
        
        print(f"Variable: {var_name}")
        print(f"KS Statistic: {ks}")
        print(f"t-Statistic: {t_statistic}, p-value: {p_value}")
        print()
```

## 1차계급화에 대해 ks통계량, t검정 2차 변수 선택

```python
from scipy.stats import ks_2samp
from statsmodels.stats.weightstats import ttest_ind

result_df = pd.DataFrame(columns=['Variable', 'KS_Statistic', 'KS_PValue',
                                  'KS_Result', 'T_Statistic', 'T_PValue',
                                  'T_Result'])
# 각 변수에 대해 KS 통계량과 t-통계량을 계산하고 결과 데이터프레임에 추가
for var in train_woe.columns[1:]:  # 첫 번째 열은 종속변수이므로 제외
    ks_stat, ks_pvalue = ks_2samp(train_woe[train_woe['위험여부'] == 0][var], train_woe[train_woe['위험여부'] == 1][var])
    
    t_stat, t_pvalue = ttest_ind(train_woe[train_woe['위험여부'] == 0][var], train_woe[train_woe['위험여부'] == 1][var])[0:2]
    # t-통계량과 p-value만 사용하도록 수정
    
    ks_result = 'yes' if ks_pvalue <= 0.05 else 'no'
    t_result = 'yes' if t_pvalue <= 0.05 else 'no'
    
    result_df = result_df.append({'Variable': var, 'KS_Statistic': ks_stat, 'KS_PValue': "{:.15f}".format(ks_pvalue),
                                  'KS_Result': ks_result, 'T_Statistic': t_stat, 'T_PValue': "{:.15f}".format(t_pvalue),
                                  'T_Result': t_result}, ignore_index=True)

# 결과 데이터프레임 출력
print(result_df)

# 결과 데이터프레임을 CSV 파일로 저장
result_df.to_csv('ks_T_statistics.csv', index=False)
```

## 스피어만 상관계수 다중공선성 확인 4차 변수 선택

```python
# 스피어만 상관계수 계산
spearman_corr = train_woe.iloc[:, 1:].corr(method='spearman')

# 스피어만 상관계수 출력 및 고상관 변수 확인
print("Spearman Correlation Matrix:")
print(spearman_corr)

# 상관계수가 0.7 이상인 변수 확인
high_corr_variables = []

threshold_corr = 0.7  # 임계값 설정

for i in range(spearman_corr.shape[0]):
    for j in range(i + 1, spearman_corr.shape[1]):
        if abs(spearman_corr.iloc[i, j]) > threshold_corr:
            high_corr_variables.append((spearman_corr.index[i], spearman_corr.columns[j], spearman_corr.iloc[i, j]))

if len(high_corr_variables) > 0:
    print("\nVariables with High Correlation (Absolute Correlation > {}):".format(threshold_corr))
    for var1, var2, corr in high_corr_variables:
        print("({}, {}): Correlation = {:.2f}".format(var1, var2, corr))
else:
    print("\nNo Variables with High Correlation")

# 고상관 변수를 CSV 파일로 저장
if len(high_corr_variables) > 0:
    corr_df = pd.DataFrame(high_corr_variables, columns=['Variable 1', 'Variable 2', 'Correlation'])
    corr_df.to_csv('high_corr_variables.csv', index=False)
```
